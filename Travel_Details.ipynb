```python
plt.figure(figsize=(12, 6))

top_destinations = df['Destination'].value_counts().head(10)

sns.barplot(
    x=top_destinations.values,
    y=top_destinations.index
)

plt.title("Top 10 Travel Destinations")
plt.xlabel("Number of Travelers")
plt.ylabel("Destination")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    df['Traveler age'],
    bins=15,
    kde=True
)

plt.title("Traveler Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    df['Duration (days)'],
    bins=15,
    kde=True
)

plt.title("Travel Duration Distribution")
plt.xlabel("Duration (Days)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Traveler gender'
)

plt.title("Traveler Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    y='Accommodation type',
    order=df['Accommodation type'].value_counts().index
)

plt.title("Accommodation Type Preferences")
plt.xlabel("Number of Travelers")
plt.ylabel("Accommodation")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

top_nationalities = df['Traveler nationality'].value_counts().head(10)

sns.barplot(
    x=top_nationalities.index,
    y=top_nationalities.values
)

plt.title("Top 10 Traveler Nationalities")
plt.xlabel("Nationality")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Traveler age',
    y='Duration (days)',
    hue='Traveler gender'
)

plt.title("Age vs Travel Duration")
plt.xlabel("Traveler Age")
plt.ylabel("Duration (Days)")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

numeric_df = df.select_dtypes(include='number')

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Travel Dataset Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='Accommodation type',
    hue='Traveler gender'
)

plt.title("Accommodation Preferences by Gender")
plt.xticks(rotation=30)
plt.show()
```

```python
destination_duration = (
    df.groupby('Destination')['Duration (days)']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=destination_duration.values,
    y=destination_duration.index
)

plt.title("Average Travel Duration by Destination")
plt.xlabel("Average Duration (Days)")
plt.ylabel("Destination")
plt.show()
```
